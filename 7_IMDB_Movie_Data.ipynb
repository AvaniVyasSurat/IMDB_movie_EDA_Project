### Data Cleaning and Preprocessing

```python
movie = movie.rename(columns= {'Revenue (Millions)' : 'revenue(millions)'})
```

This code renames the column 'Revenue (Millions)' to 'revenue(millions)' for easier and more consistent access.

# Renaming 'Runtime (Minutes)' column for easier use:

```python
movie = movie.rename(columns= {'Runtime (Minutes)' : 'runtime(minutes)'})
```

This code renames the column 'Runtime (Minutes)' to 'runtime(minutes)' for easier and more consistent access.

# Checking for missing values in each column:

```python
movie.isnull().sum()
```

This code checks for missing (null) values in each column of the `movie` DataFrame and sums them up, indicating how many missing entries are in each column.

# Checking data types of each column:

```python
movie.dtypes
```

This code displays the data type of each column in the `movie` DataFrame, which is essential for understanding how the data is stored and can be processed.

# Filling missing values in 'revenue(millions)' with 0:

```python
movie['revenue(millions)'] = movie['revenue(millions)'].fillna(0)
```

This code fills any missing values (NaN) in the 'revenue(millions)' column with `0`, ensuring no missing revenue data.

# Filling missing values in 'Metascore' with 0:

```python
movie['Metascore'] = movie['Metascore'].fillna(0)
```

This code fills any missing values (NaN) in the 'Metascore' column with `0`, ensuring no missing Metascore data.

# Re-checking for missing values after imputation:

```python
movie.isnull().sum()
```

This code re-checks for missing (null) values in each column after the imputation steps to confirm that all missing values have been handled.

### Data Analysis and Visualization

```python
movie['Director'].nunique()
```

This code counts the number of unique directors in the dataset.

# Counting movies directed by each director:

```python
top_director = movie.groupby('Director')['Rank'].count()
top_director
#top_director.idxmax()
```

This code groups the movies by 'Director' and counts the number of films each director has, storing it in `top_director`.

# Identifying the top 3 highest-rated movies by year:

```python
high_rated_movie = movie.sort_values(by = 'Rating',ascending=False)[['Title','Year']].head(3)
high_rated_movie
```

This code identifies the top 3 highest-rated movies by sorting the DataFrame by 'Rating' in descending order and selecting the 'Title' and 'Year' columns.

# Visualizing the distribution of the top 3 highest-rated movies:

```python
pl.pie(x=high_rated_movie.index ,labels=high_rated_movie.Title,autopct='%1.1f%%')
pl.show()
```

This code generates a pie chart to visualize the distribution of the top 3 highest-rated movies based on their titles and ratings.

# Finding the director with the minimum number of movies:

```python
top_director.idxmin()
```

This code finds the director who has directed the minimum number of movies based on the `top_director` Series.

# Identifying movies from the top 5 directors by number of films:

```python
top_5_directors_names = top_director.nlargest(5).index
top_5_movies = movie[movie['Director'].isin(top_5_directors_names)]
display(top_5_movies[['Title', 'Director']].head(5))
```

This code first identifies the names of the top 5 directors by the number of films they directed and then filters the main `movie` DataFrame to show movies by these top directors.

# *Visualizing* total revenue generated by the top directors:

```python
director_revenue = top_5_movies.groupby('Director')['revenue(millions)'].sum().reset_index()
pl.bar(x=director_revenue['Director'], height=director_revenue['revenue(millions)'])
pl.xlabel('Director')
pl.ylabel('Total Revenue (Millions)')
pl.title('Total Revenue by Top Directors')
pl.xticks(rotation=45, ha='right')
pl.tight_layout()
pl.show()
```

This code calculates the total revenue generated by each of the top directors and then visualizes this data using a bar chart, showing which directors have accumulated the most revenue.

# Identifying top movies by revenue:

```python
booster_movies = movie.sort_values(by = 'revenue(millions)',ascending=False)
booster_movies.head(5)
Top_5_booster_movie = booster_movies.head(5)
Top_5_booster_movie
```

This code sorts movies by 'revenue(millions)' in descending order to identify and display the top 5 movies with the highest revenue.

# Identifying top movies by Metascore:

```python
booster_movies_by_metascore = movie.sort_values(by = 'Metascore',ascending=False)[['Title','Votes']]
booster_movies_by_metascore.head(5)
```

This code sorts movies by 'Metascore' in descending order to identify and display the top 5 movies with the highest Metascore, along with their 'Title' and 'Votes'.

# Visualizing the relationship between votes and Metascore for top movies:

```python
pl.scatter(x=booster_movies_by_metascore.Votes, y=booster_movies_by_metascore.index)
pl.show()
```

This code generates a scatter plot to visualize the relationship between the number of 'Votes' and the Metascore for the top movies sorted by Metascore.

# Visualizing the revenue of the top 5 booster movies:

```python
pl.figure(figsize=(10,6))
pl.bar(x=Top_5_booster_movie['Title'], height=Top_5_booster_movie['revenue(millions)'])
pl.show()
```

This code creates a bar chart to visualize the revenue of the top 5 booster movies, allowing for a clear comparison of their financial performance.
